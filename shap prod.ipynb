```python
import numpy as np
```

```python
range = 5
X1 = np.linspace(-range, range, num=1 + 2*range, endpoint=True, retstep=True)
X2 = np.linspace(-range, range, num=1 + 2*range, endpoint=True, retstep=True)
```

```python
Ef, count = 0, 0
for x1 in X1:
    for x2 in X2:
        count += 1
        Ef += x1**2 * x2
        print(count, x1, x2, Ef)
```
